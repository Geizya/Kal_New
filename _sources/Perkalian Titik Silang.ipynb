# Perkalian Titik Silang 

## Kali Silang

Misalkan $\vec{u} = \begin{bmatrix} u_1 \\ u_2 \\ u_3 \end{bmatrix}$ dan $\vec{v} = \begin{bmatrix} v_1 \\ v_2 \\ v_3 \end{bmatrix}$ adalah vektor-vektor di $\mathbb{R}^3$.


Hasil kali silang dari $\vec{u}$ dan $\vec{v}$, dinotasikan sebagai $\vec{u} \times \vec{v}$, adalah vektor

$$
\vec{u} \times \vec{v} = \begin{bmatrix}
u_2 v_3 - u_3 v_2 \\
-(u_1 v_3 - u_3 v_1) \\
u_1 v_2 - u_2 v_1
\end{bmatrix}
$$

### Contoh Menghitung Hasil Kali Silang

Misalkan $\vec{u} = \begin{bmatrix} 2 \\ -1 \\ 4 \end{bmatrix}$ dan $\vec{v} = \begin{bmatrix} 3 \\ 2 \\ 5 \end{bmatrix}$. Temukan $\vec{u} \times \vec{v}$, dan

verifikasi bahwa hasil kali silang ini tegak lurus terhadap baik $\vec{u}$ maupun $\vec{v}$.

### Solusi

Menggunakan Definisi

$$
\vec{u} \times \vec{v} = \begin{bmatrix}
(-1)(5) - (4)(2) \\
-((2)(5) - (4)(3)) \\
(2)(2) - (-1)(3)
\end{bmatrix}
=
\begin{bmatrix}
-5 - 8 \\
-(10 - 12) \\
4 - (-3)
\end{bmatrix}
=
\begin{bmatrix}
-13 \\
-(-2) \\
7
\end{bmatrix}
=
\begin{bmatrix}
-13 \\
2 \\
7
\end{bmatrix}
$$

Cross product adalah operasi matematika antara dua vektor dalam ruang 3 dimensi (R³) yang menghasilkan vektor baru yang:

- Tegak lurus (ortogonal) terhadap kedua vektor asal.
- Memiliki arah sesuai aturan tangan kanan(right-hand rule).
- Memiliki besar (magnitudo) yang sebanding dengan luas jajar genjang yang di bentuk oleh kedua vektor dalam sesuai rumus:




$$
|a \times b| = |a| \cdot |b| \cdot \sin(\theta)
$$

Sifat sifat cross product:
- Tidak komutatif 
- Hasilnya selalu vektor bukan skalar 
- Jika dua vektor sejajar, hasil cross product-nya adalah vektor nol

1. Tentukan Luas jajarangenjang yang ditentukan oleh vector



$$
\vec{u} = \begin{bmatrix} 1 \\ 2 \end{bmatrix} \quad \text{dan} \quad \vec{v} = \begin{bmatrix} 2 \\ 1 \end{bmatrix}
$$



jawab:





Untuk menentukan luas jajar genjang yang ditentukan oleh dua vektor $\vec{u}$ dan  $\vec{v}$  dalam dua dimensi, kita dapat menggunakan nilai mutlak dari determinan matriks yang kolom-kolomnya adalah vektor-vektor tersebut.

Diberikan vektor:


$$
\vec{u} = \begin{bmatrix} 1 \\ 2 \end{bmatrix} \quad \text{dan} \quad \vec{v} = \begin{bmatrix} 2 \\ 1 \end{bmatrix}
$$

Kita dapat membentuk matriks  $A$  dengan  $\vec{u}$  sebagai kolom pertama dan  $\vec{v}$  sebagai kolom kedua:


$$
A = \begin{bmatrix} 1 & 2 \\ 2 & 1 \end{bmatrix}
$$


Luas jajar genjang  ($L$)  adalah nilai mutlak dari determinan matriks  $A$:


$$
L = |\det(A)|
$$


Hitung determinannya:


$$
\det(A) = (1)(1) - (2)(2)
$$


$$
\det(A) = 1 - 4
$$


$$
\det(A) = -3
$$


Sekarang, ambil nilai mutlaknya untuk mendapatkan luas:


$$
L = |-3|
$$
$$
L = 3
$$

Jadi, luas jajar genjang yang ditentukan oleh vektor  $\vec{u}$  dan  $\vec{v}$  adalah  $3$  satuan luas.

2. Tentukan Luas jajarangenjang yang ditentukan oleh vector



$$
\vec{u} = \begin{bmatrix} 2 \\ 0 \end{bmatrix} \quad \text{dan} \quad \vec{v} = \begin{bmatrix} 0 \\ 3 \end{bmatrix}
$$


jawab:


Untuk menentukan luas jajar genjang yang ditentukan oleh dua vektor  $\vec{u}$   dan  $\vec{v}$  dalam dua dimensi, kita dapat menggunakan nilai mutlak dari determinan matriks yang kolom-kolomnya adalah vektor-vektor tersebut.

Diberikan vektor:


$$
\vec{u} = \begin{bmatrix} 2 \\ 0 \end{bmatrix} \quad \text{dan} \quad \vec{v} = \begin{bmatrix} 0 \\ 3 \end{bmatrix}
$$


Kita dapat membentuk matriks  $A$  dengan  $\vec{u}$  sebagai kolom pertama dan  $\vec{v}$  sebagai kolom kedua:


$$
A = \begin{bmatrix} 2 & 0 \\ 0 & 3 \end{bmatrix}
$$


Luas jajar genjang  ($L$)  adalah nilai mutlak dari determinan matriks  $A$:


$$
L = |\det(A)|
$$


Hitung determinannya:


$$
\det(A) = (2)(3) - (0)(0)
$$


$$
\det(A) = 6 - 0
$$


$$
\det(A) = 6
$$


Sekarang, ambil nilai mutlaknya untuk mendapatkan luas:


$$
L = |6|
$$


$$
L = 6
$$


Jadi, luas jajar genjang yang ditentukan oleh vektor  $\vec{u}$  dan  $\vec{v}$  adalah  $6$  satuan luas.

3. Tentukan luas segitiga dengan titik titik sudut (0,0,0), (1,3,-1), dan (2,1,1).

jawab:


Untuk menentukan luas segitiga dengan titik-titik sudut di ruang 3D, kita dapat menggunakan vektor. Misalkan titik-titik sudutnya adalah  $P(0,0,0)$,  $Q(1,3,-1)$,  dan  $R(2,1,1)$.

Langkah pertama adalah membentuk dua vektor yang berasal dari salah satu titik, misalnya dari titik  $P$.

Vektor  $\vec{PQ}$  adalah vektor dari  $P$  ke  $Q$ :


$$
\vec{PQ} = Q - P = (1-0, 3-0, -1-0) = \begin{bmatrix} 1 \\ 3 \\ -1 \end{bmatrix}
$$

Vektor  $\vec{PR}$  adalah vektor dari  $P$  ke  $R$ :


$$
\vec{PR} = R - P = (2-0, 1-0, 1-0) = \begin{bmatrix} 2 \\ 1 \\ 1 \end{bmatrix}
$$


Luas segitiga adalah setengah dari besar (magnitudo) hasil perkalian silang (cross product) dari dua vektor ini:


$$
L = \frac{1}{2} ||\vec{PQ} \times \vec{PR}||
$$


Sekarang, hitung perkalian silang  $\vec{PQ} \times \vec{PR}$  :


$$
\vec{PQ} \times \vec{PR} = \begin{vmatrix} \mathbf{i} & \mathbf{j} & \mathbf{k} \\ 1 & 3 & -1 \\ 2 & 1 & 1 \end{vmatrix}
$$


$$
= \mathbf{i}((3)(1) - (-1)(1)) - \mathbf{j}((1)(1) - (-1)(2)) + \mathbf{k}((1)(1) - (3)(2))
$$


$$
= \mathbf{i}(3 + 1) - \mathbf{j}(1 + 2) + \mathbf{k}(1 - 6)
$$


$$
= 4\mathbf{i} - 3\mathbf{j} - 5\mathbf{k}
$$


Jadi, $\vec{PQ} \times \vec{PR} = \begin{bmatrix} 4 \\ -3 \\ -5 \end{bmatrix}$.

Selanjutnya, hitung besar (magnitudo) dari vektor hasil perkalian silang:


$$
||\vec{PQ} \times \vec{PR}|| = \sqrt{(4)^2 + (-3)^2 + (-5)^2}
$$


$$
= \sqrt{16 + 9 + 25}
$$


$$
= \sqrt{50}
$$


$$
= \sqrt{25 \times 2}
$$


$$
= 5\sqrt{2}
$$


Terakhir, hitung luas segitiga:


$$
L = \frac{1}{2} ||\vec{PQ} \times \vec{PR}|| = \frac{1}{2} (5\sqrt{2}) = \frac{5\sqrt{2}}{2}
$$


Jadi, luas segitiga dengan titik-titik sudut  $(0,0,0)$,  $(1,3,-1)$,  dan  $(2,1,1)$  adalah  $\frac{5\sqrt{2}}{2}$  satuan luas.

4. Tentukan luas segitiga dengan titik titik sudut (5,2,-1), (3,6,2), dan (1,0,4).

jawab:


Untuk menentukan luas segitiga dengan titik-titik sudut di ruang 3D, kita dapat menggunakan vektor. Misalkan titik-titik sudutnya adalah  $A(5,2,-1)$,  $B(3,6,2)$,  dan  $C(1,0,4)$  .

Langkah pertama adalah membentuk dua vektor yang berasal dari salah satu titik, misalnya dari titik  $A$.

Vektor

$\vec{AB}$ adalah vektor dari $A$ ke $B$:


$$
\vec{AB} = B - A = (3-5, 6-2, 2-(-1)) = \begin{bmatrix} -2 \\ 4 \\ 3 \end{bmatrix}
$$

Vektor


$\vec{AC}$ adalah vektor dari $A$ ke $C$:


$$
\vec{AC} = C - A = (1-5, 0-2, 4-(-1)) = \begin{bmatrix} -4 \\ -2 \\ 5 \end{bmatrix}
$$


Luas segitiga adalah setengah dari besar (magnitudo) hasil perkalian silang (cross product) dari dua vektor ini:


$$
L = \frac{1}{2} ||\vec{AB} \times \vec{AC}||
$$


Sekarang, hitung perkalian silang  $\vec{AB} \times \vec{AC}$ :


$$
\vec{AB} \times \vec{AC} = \begin{vmatrix} \mathbf{i} & \mathbf{j} & \mathbf{k} \\ -2 & 4 & 3 \\ -4 & -2 & 5 \end{vmatrix}
$$


$$
= \mathbf{i}((4)(5) - (3)(-2)) - \mathbf{j}((-2)(5) - (3)(-4)) + \mathbf{k}((-2)(-2) - (4)(-4))
$$


$$
= \mathbf{i}(20 + 6) - \mathbf{j}(-10 + 12) + \mathbf{k}(4 + 16)
$$


$$
= 26\mathbf{i} - 2\mathbf{j} + 20\mathbf{k}
$$


Jadi, $\vec{AB} \times \vec{AC} = \begin{bmatrix} 26 \\ -2 \\ 20 \end{bmatrix}$.

Selanjutnya, hitung besar (magnitudo) dari vektor hasil perkalian silang:


$$
||\vec{AB} \times \vec{AC}|| = \sqrt{(26)^2 + (-2)^2 + (20)^2}
$$


$$
= \sqrt{676 + 4 + 400}
$$


$$
= \sqrt{1080}
$$


Untuk menyederhanakan  $\sqrt{1080}$:


$$
\sqrt{1080} = \sqrt{36 \times 30} = 6\sqrt{30}
$$


Terakhir, hitung luas segitiga:


$$
L = \frac{1}{2} ||\vec{AB} \times \vec{AC}|| = \frac{1}{2} (6\sqrt{30}) = 3\sqrt{30}
$$


Jadi, luas segitiga dengan titik-titik sudut  $(5,2,-1)$,  $(3,6,2)$,  dan  $(1,0,4)$  adalah  $3\sqrt{30}$  satuan luas.
