### Definisi Metriks
susunan atau sekumpulan elemen (biasanya angka atau variabel) yang disusun dalam baris dan kolom yang membentuk suatu tabel persegi panjang atau kotak. Dalam notasi matematika, matriks sering kali dituliskan dalam bentuk huruf kapital seperti
A,B dan M

### Konsep dasar matriks
Notasi matriks biasanya ditulis dalam huruf kapital tebal seperti A, B, C, dll, dan elemen-elemen dalam matriks ditulis dalam tanda kurung atau kurung siku

### Operasi Aritmatika Metriks
Operasi aritmatika matriks adalah proses perhitungan matematika yang melibatkan satu atau lebih matriks, di mana setiap elemen dalam matriks dihitung sesuai dengan aturan operasi matematika dasar, seperti penjumlahan, pengurangan, perkalian, dan pembagian (meskipun pembagian langsung jarang digunakan pada matriks).
#### Syarat aritmatika metriks
1.Penjumlahan dan Pengurangan: Ukuran matriks harus sama.
2.Perkalian Matriks Jumlah kolom matriks pertama harus sama dengan jumlah baris matriks kedua.

```python
##CODE PENJUMLAHAN METRIKS

A = [
    [3, 1],
    [5, 4]
]
B = [
    [7, 6],
    [2, 3]
]
result = [
    [0, 0],
    [0, 0]
]
for i in range(len(A)):
    for j in range(len(A[0])):
        result[i][j] = A[i][j] + B[i][j]
print("Hasil penjumlahan matriks A + B:")
for row in result:
    print(row)
```

```python
##CODE PENGURANGAN METRIKS

A = [
    [7, 4],
    [3, 8]
]

B = [
    [2, 3],
    [4, 5]
]

hasil = [
    [0, 0],
    [0, 0]
]

for i in range(2):        # 2 baris
    for j in range(2):    # 2 kolom
        hasil[i][j] = A[i][j] - B[i][j]

print("Hasil Pengurangan Matriks A - B:")
for row in hasil:
    print(row)
```

```python
##CODE PERKALIAN METRIKS
import numpy as np

A = np.array([[1, 0], [0, 1]])
B = np.array([[5, 6], [7, 8]])
C = np.array([[1, 1], [0, 1]])

# Perkalian A x B x C
result = A @ B @ C

print("Hasil Perkalian A x B x C:\n", result)
```

```python
## Perkalian 2 metriks atau lebih
def perkalian_matriks(A, B):
    # Ukuran: A(m x n), B(n x p)
    m = len(A)
    n = len(A[0])
    p = len(B[0])

    hasil = [[0 for _ in range(p)] for _ in range(m)]

    for i in range(m):
        for j in range(p):
            for k in range(n):
                hasil[i][j] += A[i][k] * B[k][j]

    return hasil

A = [
    [1, 0],
    [0, 1]
]

B = [
    [2, 3],
    [4, 5]
]

C = [
    [1, 1],
    [0, 1]
]

# Perkalian pertama: A x B
hasil1 = perkalian_matriks(A, B)

# Perkalian kedua: hasil1 x C
hasil2 = perkalian_matriks(hasil1, C)

print("Hasil A x B:")
for baris in hasil1:
    print(baris)

print("\nHasil (A x B) x C:")
for baris in hasil2:
    print(baris)
```

```python
## Perkalian skalar
M = [
    [1, 2],
    [3, 4]
]

# Skalar
k = 3

# Perkalian skalar ke setiap elemen matriks
hasil = [[k * elemen for elemen in baris] for baris in M]

# Tampilkan hasil
print("Hasil perkalian skalar:")
for row in hasil:
    print(row)
```

## Determinan Metriks
Merupakan suatu nilai skalar yang dapat dihitung dari elemen-elemen suatu matriks persegi yang biasanya matriks dengan jumlah baris dan kolom yang sama.
digunakan untuk mementukan apakah suatu matriks memiliki invers (det (A)=/ 0).

 * sifat sifat penting determinan
1. Determinan Matriks identitas:
det(I)=1, dimana I adalah matriks identitas
2. Baris atau kolom nol:
Jika ada baris atau kolom yang semuanya nol, maka det(A)=0
3. pertukaran baris/kolom;
Jika dua baris atau kolom dipertukarkan, determinan berubah tanda
4. Kelipatan baris/kolom:
Jika satu baris atau kolom dikalikan dengan skalar k, maka determinan menjadi k det (A)
5. matriks singular:
Jika det(A)=0, maka matriks A disebut singular(Tidak memiliki invers) jika non singular maka matriks nya buka 0, tapi -1 memiliki invers (sebalik nya dari matriks singular)

* contoh matriks 2x2:

$$
\begin{bmatrix}
& a & b &  \\
& c & d & \\
\end{bmatrix}
\quad
$$

maka determinannya adalah:
det(A) = ad−bc

* contoh matriks 3x3:

$$
\begin{bmatrix}
& a & b & c & \\
& d & e & f & \\
& g & h & i & \\
\end{bmatrix}
\quad
$$

maka determinannya adalah:
det(A) = a(ei−fh)−b(di−fg)+c(dh−eg)




## Kofaktor matriks
Merupakan kofaktor dari suatu elemen dalam matriks adalah minor dari elemen tersebut yang dikalikan dengan faktor tanda berdasarkan posisinya dalam matriks.

* Rumus Kofaktor

$$
C_{ij} = (-1)^{i+j} M_{ij}
$$
* contoh matriks 3x3:
* menentukan minor

A =

$$
\begin{bmatrix}
& 1 & 2 & 3 &\\
& 4 & 5 & 6 &\\
& 7 & 8 & 9 &\\
\end{bmatrix}
\quad
$$

Minor untuk elemen $A_{23}$ = 6

$
M_{23}:
$
$$
\begin{bmatrix}
& 1 & 2 &  \\
& 7 & 8 & \\
\end{bmatrix}
\quad
$$

$M_{23}$ ​= (1×8)−(2×7) = 8−14 = −6
* lalu masukkan rumus kofaktor

$$
C_{ij} = (-1)^{i+j} M_{ij}
$$
* masukkan angka
$$
C_{23} = (-1)^{2+3} (-6)
$$
* Karena 2+3 = 5, maka $(−1)^{5}$ = −1, sehingga:

$$
C_{23} = (-1) × (-6) = 6
$$

#### Hubungan Matriks Minor & Kofaktor
* Minor $(M_{ij})$ adalah determinan dari submatriks yang diperoleh dengan menghapus baris ke-i dan kolom ke-j.
* Kofaktor $(C_{ij})$ adalah minor yang dikalikan dengan faktor tanda $(−1)^{i+j}$ dan digunakan dalam perhitungan determinan, invers matriks, serta ekspansi kofaktor.
