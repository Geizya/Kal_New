# Perkalian Titik (Dot Product)
##  1. Perkalian Titik
Dalam bagian ini, kita memperkenalkan operasi aljabar sederhana yang dikenal sebagai hasil kali titik (dot product), yang membantu kita mengukur panjang vektor dan sudut yang dibentuk oleh sepasang vektor. Untuk vektor-vektor dua dimensi v dan w, hasil kali titik (v · w) didefinisikan sebagai bilangan skalar berikut:

$\mathbf{v} \cdot \mathbf{w} =
\begin{bmatrix}
v_1 \\
v_2
\end{bmatrix}
\cdot
\begin{bmatrix}
w_1 \\
w_2
\end{bmatrix}
= v_1w_1 + v_2w_2.$

Sebagai contoh,

$\begin{bmatrix}
2 \\
-3
\end{bmatrix}
\cdot
\begin{bmatrix}
4 \\
1
\end{bmatrix}
= 2 \cdot 4 + (-3) \cdot 1 = 5.$

berikut program untuk menghitung dit product (perkalian tiitk dari dua vektor)

```python
def dot_product(v1, v2):
    if len(v1) != len(v2):
        raise ValueError("Panjang kedua vektor harus sama")

    result = sum(a * b for a, b in zip(v1, v2))
    return result

# Contoh penggunaan
vec1 = [2, -3]
vec2 = [4, 1]

hasil = dot_product(vec1, vec2)
print("Dot product:", hasil)
```

## 2. Vektor satuan $\mathbb{R}^n$
Vektor satuan (unit vector) adalah vektor dengan panjang satu.

Jelas bahwa vektor x adalah vektor satuan jika dan hanya jika:

$\|\mathbf{x}\|^2 = \mathbf{x} \cdot \mathbf{x} = 1.$

### Membentuk Vektor Satuan
Misalkan x adalah vektor tak nol di $\mathbb{R}^n$ . Maka:

$\frac{\mathbf{x}}{\|\mathbf{x}\|}$

Bukti:

$\frac{\mathbf{x}}{\|\mathbf{x}\|} \cdot \frac{\mathbf{x}}{\|\mathbf{x}\|} = \frac{\mathbf{x} \cdot \mathbf{x}}{\|\mathbf{x}\|^2} = \frac{\|\mathbf{x}\|^2}{\|\mathbf{x}\|^2} = 1$

#### Contoh :    
Misal $\mathbf{x} = \begin{bmatrix} 3 \\ 4 \end{bmatrix}$:

- Panjang Vektor =
$\|\mathbf{x}\| = \sqrt{3^2 + 4^2} = \sqrt{9 + 16} = \sqrt{25} = 5.$

- Vektor satuan =
$\mathbf{u} = \frac{\mathbf{x}}{\|\mathbf{x}\|} = \frac{1}{5} \begin{bmatrix} 3 \\ 4 \end{bmatrix} = \begin{bmatrix} \frac{3}{5} \\ \frac{4}{5} \end{bmatrix}.$

- Verifikasi =
$\|\mathbf{u}\| = \sqrt{\left(\frac{3}{5}\right)^2 + \left(\frac{4}{5}\right)^2} = \sqrt{\frac{9}{25} + \frac{16}{25}} = \sqrt{1} = 1.$

berikut adalah program for mengkonversi vektor menjadi vektor satuan (vektor memiliki panjang vektor 1)

```python
import math

def hitung_panjang_vektor(v):
    return math.sqrt(sum([x**2 for x in v]))

def vektor_satuan(v):
    panjang = hitung_panjang_vektor(v)
    if panjang == 0:
        raise ValueError("Vektor nol tidak bisa dikonversi ke vektor satuan.")
    return [x / panjang for x in v]

# Contoh penggunaan:
x = [3, 4]
u = vektor_satuan(x)

print("Vektor satuan:", u)

# Verifikasi panjang vektor satuan
panjang_u = hitung_panjang_vektor(u)
print("Panjang vektor satuan:", panjang_u)
```
