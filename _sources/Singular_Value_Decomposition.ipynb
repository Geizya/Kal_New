# Singular Value Decomposition

## Apa itu SVD
Singular Value Decomposition (SVD) atau Dekomposisi Nilai Singular adalah teknik aljabar linear fundamental yang banyak digunakan dalam berbagai bidang seperti ilmu data, machine learning, pemrosesan sinyal, statistik, dan komputasi numerik.

Secara sederhana, SVD adalah cara untuk memecah (mendekomposisi) sebuah matriks menjadi tiga matriks yang lebih sederhana dan bermakna. Ini mirip dengan bagaimana Anda bisa memfaktorkan angka (misalnya, 12 = 2 x 2 x 3), tetapi untuk matriks.

#### Rumus matematis SVD:
Untuk matriks $A$ berukuran $m×n$, dekomposisinya adalah:
$A = U \Sigma V^T$

**Dimana:**

- $A$: Matriks asli yang ingin Anda dekomposisi $(m×n)$.
- $U$: Matriks ortogonal kiri $(m×m)$. Kolom-kolom U disebut vektor singular kiri (left singular vectors).
- $Σ$ (Sigma): Matriks diagonal $(m×n)$ yang berisi nilai-nilai singular (singular values) matriks A. Nilai-nilai singular ini biasanya diurutkan dari yang terbesar ke yang terkecil. Nilai-nilai di luar diagonal adalah nol.
- $V^T$: Transpose dari matriks ortogonal kanan $(n×n)$. Kolom-kolom V (atau baris-baris $V^T$) disebut vektor singular kanan (right singular vectors).

## Kegunaan SVD
SVD memiliki banyak kegunaan, termasuk dalam pengurangan dimensi, ekstraksi fitur, kompresi data, dan analisis data. Dalam pengolahan citra, SVD digunakan untuk kompresi dan rekonstruksi gambar, serta untuk pemulihan citra yang terdegradasi.
kegunaan utama dari SVD (Singular Value Decomposition) dalam berbagai bidang:
✅ Pengurangan Dimensi : menyederhanakan data tanpa kehilangan informasi penting (dipakai di PCA).
🖼️ Kompresi Gambar : mengurangi ukuran gambar dengan tetap mempertahankan kualitas visual.
🤖 Machine Learning & NLP : memahami hubungan antar kata/dokumen (misalnya LSA).
🎯 Sistem Rekomendasi : memprediksi preferensi pengguna (misalnya di Netflix, YouTube).
🧮 Solusi Sistem Linear : menyelesaikan sistem persamaan tak lengkap atau tak konsisten.
🔍 Deteksi Anomali : mengenali pola yang tidak biasa dalam data.


Berikut adalah beberapa kegunaan SVD secara lebih detail:

#### 1. Pengurangan Dimensi:
- SVD dapat digunakan untuk mengurangi jumlah variabel dalam dataset tanpa kehilangan banyak informasi. Ini sangat berguna dalam dataset berdimensi tinggi, di mana pemrosesan data bisa menjadi rumit.
- Dalam konteks ini, SVD membantu mengidentifikasi fitur-fitur penting dan membuang yang tidak penting, sehingga mempercepat proses komputasi dan analisis.
#### 2. Ekstraksi Fitur:
- SVD membantu mengekstrak fitur-fitur utama dari data. Dalam pengolahan citra, misalnya, SVD dapat mengidentifikasi pola-pola penting dalam gambar, yang kemudian dapat digunakan untuk pengenalan wajah atau objek.
- SVD juga dapat digunakan dalam pemrosesan bahasa alami untuk mengekstrak makna atau konsep yang mendasari dari teks.
#### 3. Kompresi Data:
- SVD memungkinkan kompresi data dengan mempertahankan informasi yang paling penting. Dalam pengolahan citra, SVD dapat digunakan untuk mengompresi gambar dengan menghilangkan nilai singular yang kecil, yang berkontribusi sedikit pada kualitas gambar.
- Ini memungkinkan penyimpanan dan transmisi data yang lebih efisien tanpa kehilangan kualitas yang signifikan.
#### 4. Analisis Data:
- SVD membantu dalam analisis data dengan mengidentifikasi struktur dan hubungan yang tersembunyi dalam data.
- Dalam analisis data, SVD dapat digunakan untuk menemukan korelasi antar variabel dan mengidentifikasi kelompok data yang serupa.
#### 5. Rekomendasi:
- SVD digunakan dalam sistem rekomendasi untuk memprediksi preferensi pengguna berdasarkan riwayat interaksi mereka dengan item.
- Dengan menguraikan matriks preferensi pengguna-item, SVD dapat mengidentifikasi pola dan rekomendasi yang dipersonalisasi.
#### 6. Pemulihan Citra yang Terdegradasi:
- SVD dapat digunakan untuk memulihkan citra yang mengalami kerusakan, seperti piksel yang hilang.
- Dengan menggunakan SVD, nilai-nilai piksel yang hilang dapat diisi kembali dengan nilai yang diperkirakan dari matriks yang direkonstruksi.
#### 7. Pengolahan Sinyal:
- SVD membantu dalam menyaring sinyal dan menghilangkan gangguan. Ini berguna dalam berbagai aplikasi pengolahan sinyal, seperti dalam analisis data seismik atau audio.
#### 8. Pemeringkatan Efektif Matriks:
- SVD dapat digunakan untuk menentukan peringkat efektif suatu matriks, terutama dalam kasus di mana kesalahan pembulatan dapat menyebabkan nilai singular yang kecil tetapi bukan nol.

### Contoh Soal 1

#### Matriks Awal

$$
A = \begin{bmatrix}
3 & 1 \\
1 & 3
\end{bmatrix}
$$

#### Mencari Transpose A

Transpose = baris jadi kolom:

$$
A^T = \begin{bmatrix}
3 & 1 \\
1 & 3
\end{bmatrix}
$$

Karena $A$ simetris, maka $A^T = A$.

#### Hitung $A^T A$


$$A^T A = A A =
\begin{bmatrix} 3 & 1 \\ 1 & 3 \end{bmatrix}
\begin{bmatrix} 3 & 1 \\ 1 & 3 \end{bmatrix}
=\begin{bmatrix}
10 & 6 \\
6 & 10
\end{bmatrix}$$

#### Cari Nilai Eigen $A^T A$

$$
\det(A^T A - \lambda I) = 0
$$

$$
\det \begin{bmatrix}
10 - \lambda & 6 \\
6 & 10 - \lambda
\end{bmatrix} = 0
$$

$$
(10 - \lambda)^2 - 36 = 0
$$

$$
(10 - \lambda)^2 = 36
$$

$$
10 - \lambda = \pm 6
$$

$$
\lambda = 10 \mp 6
$$

$$
\Rightarrow \lambda_1 = 16, \quad \lambda_2 = 4
$$

#### Singular Value

$$
\sigma_1 = \sqrt{16} = 4, \quad \sigma_2 = \sqrt{4} = 2
$$

#### Cari Vektor Eigen untuk $V$

Untuk $\lambda = 16$:

$$
\begin{bmatrix} -6 & 6 \\ 6 & -6 \end{bmatrix}
\Rightarrow v_1 = v_2
$$

Normalisasi:

$$
v_1 = \frac{1}{\sqrt{2}}, \quad v_2 = \frac{1}{\sqrt{2}}
$$

Untuk $\lambda = 4$:

$$
\begin{bmatrix} 6 & 6 \\ 6 & 6 \end{bmatrix}
\Rightarrow v_1 = -v_2
$$

Normalisasi:

$$
v_1 = \frac{1}{\sqrt{2}}, \quad v_2 = -\frac{1}{\sqrt{2}}
$$

#### Matriks V

$$
V = \begin{bmatrix}
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} \\
\frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{2}}
\end{bmatrix}
$$

#### Cari U

$$
U = A V \Sigma^{-1}
$$

$$
A V =
\begin{bmatrix}
2 \sqrt{2} & \sqrt{2} \\
2 \sqrt{2} & -\sqrt{2}
\end{bmatrix}
$$

Normalisasi:

$$
U = \begin{bmatrix}
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} \\
\frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{2}}
\end{bmatrix}
$$

#### Hasil Akhir SVD

$$
A = U \Sigma V^T
$$

dengan:

$$
U = \begin{bmatrix}
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} \\
\frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{2}}
\end{bmatrix}, \quad
\Sigma = \begin{bmatrix}
4 & 0 \\
0 & 2
\end{bmatrix}, \quad
V = \begin{bmatrix}
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} \\
\frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{2}}
\end{bmatrix}
$$

### Pembuktian  $A$

$A = \begin{bmatrix} 3 & 1 \\ 1 & 3 \end{bmatrix}$

$U = \frac{1}{\sqrt{2}}
\begin{bmatrix}
1 & 1 \\
1 & -1
\end{bmatrix}$

$\Sigma =
\begin{bmatrix}
4 & 0 \\
0 & 2
\end{bmatrix}$

$V = \frac{1}{\sqrt{2}}
\begin{bmatrix}
1 & 1 \\
1 & -1
\end{bmatrix}$

---

#### Karena $V\) ortogonal, \(V^T = V$

Jadi:

$$
V^T = V
$$

---

#### Hitung $\Sigma V^T)$

$$
\Sigma V^T =
\Sigma V =
\begin{bmatrix} 4 & 0 \\ 0 & 2 \end{bmatrix}
\frac{1}{\sqrt{2}}
\begin{bmatrix} 1 & 1 \\ 1 & -1 \end{bmatrix}
$$

Hasil:

$$
= \frac{1}{\sqrt{2}}
\begin{bmatrix} 4 \cdot 1 + 0 \cdot 1 & 4 \cdot 1 + 0 \cdot (-1) \\
0 \cdot 1 + 2 \cdot 1 & 0 \cdot 1 + 2 \cdot (-1)
\end{bmatrix}
= \frac{1}{\sqrt{2}}
\begin{bmatrix} 4 & 4 \\ 2 & -2 \end{bmatrix}
$$

---

#### Hitung $U \Sigma V^T$

$$
U \Sigma V^T =
\frac{1}{\sqrt{2}}
\begin{bmatrix} 1 & 1 \\ 1 & -1 \end{bmatrix}
\frac{1}{\sqrt{2}}
\begin{bmatrix} 4 & 4 \\ 2 & -2 \end{bmatrix}
$$

Karena
$$
\frac{1}{\sqrt{2}} \times \frac{1}{\sqrt{2}} = \frac{1}{2} = \frac{1}{2}
\begin{bmatrix}
1 \cdot 4 + 1 \cdot 2 & 1 \cdot 4 + 1 \cdot (-2) \\
1 \cdot 4 + (-1) \cdot 2 & 1 \cdot 4 + (-1) \cdot (-2)
\end{bmatrix}
= \frac{1}{2}
\begin{bmatrix}
6 & 2 \\
2 & 6
\end{bmatrix}
= \begin{bmatrix}
3 & 1 \\
1 & 3
\end{bmatrix} = A$$

---

#### Kesimpulan

terbukti bahwa :

$$
A = U \Sigma V^T
$$

dengan hasil sesuai matriks aslinya.

### Contoh Soal 2

#### Perhitungan Singular Value Decomposition (SVD)

 Diketahui Matriks

$$
A = \begin{bmatrix}
3 & 7 \\
2 & 5 \\
4 & 3 \\
1 & 1
\end{bmatrix}
$$

### Langkah 1: Hitung \(A^T A\)

$$A^T =\begin{bmatrix}
3 & 2 & 4 & 1 \\
7 & 5 & 3 & 1
\end{bmatrix}$$

$$A^T A =\begin{bmatrix}
3 & 2 & 4 & 1 \\
7 & 5 & 3 & 1\end{bmatrix}
\begin{bmatrix}
3 & 7 \\
2 & 5 \\
4 & 3 \\
1 & 1
\end{bmatrix}=
\begin{bmatrix}
30 & 44 \\
44 & 84
\end{bmatrix}$$

#### Langkah 2: Hitung Eigenvalue dari \(A^T A\)

Misal  $\lambda \text{ adalah eigenvalue dari } A^T A$

$$
\left|
\begin{array}{cc}
30 - \lambda & 44 \\
44 & 84 - \lambda
\end{array}
\right|
= 0
\Rightarrow
(30 - \lambda)(84 - \lambda) - 44^2 = 0
$$

$$
\lambda^2 - 114 \lambda + 2520 - 1936 = 0
\Rightarrow
\lambda^2 - 114 \lambda + 584 = 0
$$

$$
\lambda = \frac{114 \pm \sqrt{114^2 - 4 \cdot 584}}{2}
= \frac{114 \pm \sqrt{10660}}{2}
\Rightarrow
\lambda_1 \approx 108.62, \quad \lambda_2 \approx 5.38
$$

#### Langkah 3: Hitung Singular Value

$$
\sigma_1 = \sqrt{108.62} \approx 10.42, \quad
\sigma_2 = \sqrt{5.38} \approx 2.32
$$

$$
\Sigma =
\begin{bmatrix}
10.42 & 0 \\
0 & 2.32 \\
0 & 0 \\
0 & 0
\end{bmatrix}
$$

#### Langkah 4: Cari Eigenvector dari \(A^T A\)

Hasil eigenvector ter-normalisasi (dari komputasi numerik):

$$
v_1 =
\begin{bmatrix}
0.4884 \\
0.8726
\end{bmatrix}, \quad
v_2 =
\begin{bmatrix}
-0.8726 \\
0.4884
\end{bmatrix}
$$

$$
V =
\begin{bmatrix}
0.4884 & -0.8726 \\
0.8726 & 0.4884
\end{bmatrix}
$$

#### Langkah 5: Hitung Matriks \(U\)

$$
u_1 = \frac{1}{10.42} A v_1, \quad
u_2 = \frac{1}{2.32} A v_2
$$

$$
u_1 \approx
\begin{bmatrix}
0.6817 \\
0.4992 \\
0.4725 \\
0.1668
\end{bmatrix}, \quad
u_2 \approx
\begin{bmatrix}
0.3241 \\
0.3882 \\
-0.3541 \\
-0.7987
\end{bmatrix}
$$

Lengkapi \(U\) menjadi matriks ortogonal \(4 \times 4\) (dengan Gram-Schmidt atau QR decomposition):

$$
U =
\begin{bmatrix}
\vec{u}_1 & \vec{u}_2 & \vec{u}_3 & \vec{u}_4
\end{bmatrix}
$$

#### Langkah 6: Bentuk SVD

$$
A = U \Sigma V^T
$$

Dengan:

$$
\Sigma =
\begin{bmatrix}
10.42 & 0 \\
0 & 2.32 \\
0 & 0 \\
0 & 0
\end{bmatrix}, \quad
V^T =
\begin{bmatrix}
0.4884 & 0.8726 \\
-0.8726 & 0.4884
\end{bmatrix}
$$

### Pembuktian

$$
U \Sigma V^T = A
$$

Perkalian kembali ketiga matriks akan menghasilkan matriks semula (atau sangat mendekati secara numerik karena pembulatan).
